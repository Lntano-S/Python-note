## 集合类
### 6.1 集合

1. 哈希运算 `hash()`：通过元素对应的哈希值，在集合中快速对应
2. 创建集合：**`{}`** or **`set()`**, *与列表不同，其元素没有位置和序列，可以动态增加和删除*，即 *唯一性(Uniqueness),无序性(Unordered),元素不可哈希(Unhashable)*

````python
s1 = {1,2,[2,4]}    #对于可变对象(list, dict),Python 在内存中为它分配了一块空间,所以是 unhashable
s2 = {1,2,{2,4}}    #对于不可变对象(int(),str(),tuple()),这些都是可哈希的
````

3. 集合操作：
    | 操作函数or方法 | 作用 |
    |:---|:---|
    | `s.add(x)` | 添加元素 `x` |
    | `s.remove(x)` | 移除元素 `x`，如果元素不存在则报错 |
    | `s.discard(x)` | 移除元素 `x`，如果元素不存在**不**报错 |
    | `s.pop()` | 随机移除并返回集合 `s` 中的一个元素 |
    | `s.clear()` | 清空所有元素 |
    | `len(s)` | 计算集合 `s` 中元素的个数 |
    | `x in s` | 判断元素 `x` 是否在集合 `s` 中 |
    | `s.copy()` | 返回集合 `s` 的一个浅拷贝 |
    | `s.update(t)` | 用集合 `t` 中的元素更新集合 `s` |

4. 集合逻辑运算

假设有两个集合 `A = {1, 2, 3}` 和 `B = {3, 4, 5}`。

| 运算  | 方法 | 描述 | 示例结果 |
|:---|:---|:---|:---|
| **交集** | `A & B` | 返回 A 和 B 中**共有**的元素 | `{3}` |
| **并集** | `A | B` | 返回 A 和 B 中**所有**的元素（去重） | `{1, 2, 3, 4, 5}` |
| **差集** | `A - B` | 返回在 A 中但**不在 B 中**的元素 | `{1, 2}` |
| **对称差集** | `A ^ B` | 返回**只在 A 或 B 中**出现的元素（并集减去交集） | `{1, 2, 4, 5}` |

---

#### 增强赋值运算符（原地修改集合）

除了上面的运算符会返回一个**新集合**外，还有一组增强赋值运算符，它们会**直接修改**调用它们的集合（例如，修改集合 `A` 本身），并且效率更高。

| 运算 | 运算符 |
|:---|:---|
| 交集更新 | `A &= B` |
| 并集更新 | `A |= B` |
| 差集更新 | `A -= B` |
| 对称差集更新 | `A ^= B` |

---

#### 子集与超集判断

| 运算 | 运算符 | 描述 |
|:---|:---|:---|
| 子集判断 | `A <= B` | 判断 A 是否为 B 的子集 |
| 真子集判断 | `A < B` | 判断 A 是否为 B 的真子集（A 是 B 的子集，且 A 不等于 B） |
| 超集判断 | `A >= B` | 判断 A 是否为 B 的超集 |
| 真超集判断 | `A > B` | 判断 A 是否为 B 的真超集（A 是 B 的超集，且 A 不等于 B） |

````python
# exercise 1
s ={10,20,30}
s.add(20)   #集合中已有元素20，唯一性
s.discard(40)   #尝试移除不存在的元素40，不会报错
s.pop() 
print(s)
````

### 6.4 字典
> 元素是键值对的**集合**，性质类似

1. **映射关系**：`键值对`，结构为 `key : value` (索引不一定是数字)
2. **字典创建**：0个或多个键值对的集合，用 `{}` or `dict()` 创建
3. **字典操作**

#### 基本操作

| 操作 | 描述 | 示例 (假设 `d = {'a': 1, 'b': 2}`) |
|:---|:---|:---|
| `d[key]` | 通过键访问值，键不存在时会报错 | `d['a']` → `1` |
| `d.get(key)` | 通过键访问值，键不存在时返回 `None` | `d.get('c')` → `None` |
| `d.get(key, default)` | 通过键访问值，键不存在时返回指定的默认值 | `d.get('c', 0)` → `0` |
| `d[key] = value` | 添加新键值对，或修改已存在的键的值 | `d['c'] = 3` |
| `key in d` | 判断键是否在字典中 | `'a' in d` → `True` |
| `len(d)` | 返回字典中键值对的个数 | `len(d)` → `2` |
| `del d[key]` | 删除指定键的键值对 | `del d['a']` |

#### 常用方法

| 方法 | 描述 |
|:---|:---|
| `d.keys()` | 返回所有键的视图对象 |
| `d.values()` | 返回所有值的视图对象 |
| `d.items()` | 返回所有键值对（元组形式）的视图对象 |
| `d.pop(key)` | 删除指定键并返回其值，键不存在时报错 |
| `d.pop(key, default)` | 删除指定键并返回其值，键不存在时返回默认值 |
| `d.popitem()` | 随机删除并返回一个键值对（元组形式） |
| `d.clear()` | 清空字典中的所有键值对 |
| `d.update(other)` | 用另一个字典更新当前字典（添加或覆盖） |
| `d.setdefault(key, default)` | 如果键存在则返回其值；否则插入键并设为默认值 |
| `d.copy()` | 返回字典的浅拷贝 |

4. **字典应用场景**：映射表达和个数统计

````python

def add_to_cart(cart, product, price, quantity):
    # 如果商品已经在购物车里
    if product in cart:
        # 数量累加
        cart[product]['quantity'] += quantity
    else:
        # 如果商品不在购物车里
        cart[product] = {
            'price': price,
            'quantity': quantity
        }

def checkout(cart):
    total = 0 
    for product, info in cart.items():
        # 每种商品的小计
        subtotal = info['price'] * info['quantity']
        total += subtotal
    return total


if __name__ == "__main__":
    shopping_cart = {}
    
    # 第一次购买：苹果5元每斤，买3斤
    add_to_cart(shopping_cart, "苹果", 5.0, 3)
    print("加入3斤苹果到购物车")
    
    # 第二次购买：香蕉3.5元每斤，买2斤
    add_to_cart(shopping_cart, "香蕉", 3.5, 2)
    print("加入2斤香蕉到购物车")
    
    # 第三次购买：再加2斤苹果
    add_to_cart(shopping_cart, "苹果", 5.0, 2)
    print("再加2斤苹果到购物车")
    print("-" * 40)
    
    # 打印
    print("购物小票")
    print("商品名称   单价    数量    小计")
    print("-" * 40)
    
    # 计算每种商品的小计和总金额
    total_amount = 0
    
    # 遍历购物车，打印每种商品信息
    for product, info in shopping_cart.items():
        price = info['price']
        quantity = info['quantity']
        subtotal = price * quantity
        total_amount += subtotal
        
        # 格式化打印，确保对齐
        print(f"{product:6}   {price:5.1f}元   {quantity:3}    {subtotal:6.1f}元")
    
    print("-" * 40)
    print(f"总金额：{total_amount:.1f}元")
````
