```python
import matplotlib.pyplot as plt
import numpy as np
from numpy.typing import NDArray
import csv

def main() -> None:
    plt.rcParams['font.sans-serif'] = ['SimHei'] 
    plt.rcParams['axes.unicode_minus'] = False
    
    
    with open("../../Python Experiment/data1.csv", "r", encoding="utf-8") as file:
        reader = csv.reader(file)
        name: list[str] = next(reader)
        rows: list[list] = list(reader)
        
    movie1: list[int] = []
    movie2: list[int] = []
    movie3: list[int] = []
    
    for row in rows:
        movie1.append(int(row[1]))
        movie2.append(int(row[2]))
        movie3.append(int(row[3]))
        
    m1: NDArray[np.int64] = np.array(movie1)
    m2: NDArray[np.int64] = np.array(movie2)
    m3: NDArray[np.int64] = np.array(movie3)
    
    x = np.arange(len(movie1))
    width = 0.25
    
    fig, axes = plt.subplots(2, 2, figsize=(15, 10))
    
    ax1 = axes[0, 0]
    ax1.bar(x - width, m1, width=width, label=name[1])
    ax1.bar(x, m2, width=width, label=name[2])
    ax1.bar(x + width, m3, width=width, label=name[3])
    ax1.set_title('电影数据对比')
    ax1.set_xlabel('时间')
    ax1.set_ylabel('数值')
    ax1.legend()
    
    ax2 = axes[0, 1]
    ax2.bar(x, m1, label=name[1])
    ax2.bar(x, m2, bottom=m1, label=name[2])
    ax2.bar(x, m3, bottom=m1+m2, label=name[3])
    ax2.set_title('-电影数据堆叠对比')
    ax2.set_xlabel('时间')
    ax2.set_ylabel('数值')
    ax2.legend()
    
    ax3 = axes[1, 0]
    # 取第三天的数据 (索引为2)
    pie_data = [m1[2], m2[2], m3[2]]
    pie_labels = [name[1], name[2], name[3]]
    ax3.pie(pie_data, labels=pie_labels, autopct='%1.1f%%')
    ax3.set_title('三部电影第三天票房占比')
    
    ax4 = axes[1, 1]
    ax4.plot(x, m1, label=name[1], marker='o')
    ax4.plot(x, m2, label=name[2], marker='s')
    ax4.plot(x, m3, label=name[3], marker='^')
    ax4.set_title('电影票房趋势折线图')
    ax4.set_xlabel('时间')
    ax4.set_ylabel('数值')
    ax4.grid(True)
    ax4.legend()
    

    plt.tight_layout()  
    plt.savefig('analysis.png')  
    plt.show()
    
    _list: list[int] = [1, 3, 4]
    
if __name__ == "__main__":
    main()
```
